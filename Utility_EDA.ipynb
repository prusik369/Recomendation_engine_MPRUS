```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Plot style settings
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
```

```python
# counts missing values for each column and prints a clean report
def check_missing_values(df):
    missing = df.isna().sum()
    print("Missing values per column:")
    print(missing)
    return missing
```

```python
# prints the min and max timestamp in the df
def show_date_range(df, date_col):
    # make sure the column is datetime
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    print("Date range:")
    print(df[date_col].min(), " ---> ", df[date_col].max())
```

```python
# draws a histogram for a numeric column
def plot_histogram(df, col, bins=30):
    plt.figure(figsize=(8, 4))
    plt.hist(df[col].dropna(), bins=bins)
    plt.title(f"Histogram for {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
# draws a time series for a column based on datetime
def plot_time_series(df, date_col, value_col):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    df = df.sort_values(date_col)

    plt.figure(figsize=(12, 5))
    plt.plot(df[date_col], df[value_col])
    plt.title(f"Time series of {value_col}")
    plt.xlabel(date_col)
    plt.ylabel(value_col)
    plt.show()
```

```python
# histogram aggregated by a time unit: day, week, month, year
def time_unit_histogram(df, date_col, unit):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

    if unit == "day":
        grouped = df.groupby(df[date_col].dt.day).size()
    elif unit == "week":
        grouped = df.groupby(df[date_col].dt.isocalendar().week).size()
    elif unit == "month":
        grouped = df.groupby(df[date_col].dt.month).size()
    elif unit == "year":
        grouped = df.groupby(df[date_col].dt.year).size()
    else:
        raise ValueError("unit must be one of: day, week, month, year")

    plt.figure(figsize=(8, 4))
    plt.bar(grouped.index, grouped.values)
    plt.title(f"Histogram grouped by {unit}")
    plt.xlabel(unit)
    plt.ylabel("Count")
    plt.show()
```
